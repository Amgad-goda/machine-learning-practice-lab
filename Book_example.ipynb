```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
from numpy._core.fromnumeric import sort
# Download and prepare the data
data_root = "https://github.com/ageron/data/raw/main/"
lifesat = pd.read_csv(data_root + "lifesat/lifesat.csv")

#dfNew = lifesat.sort_values(by="Life satisfaction", ascending=False)
#print(dfNew)

x, y = lifesat[["GDP per capita (USD)"]], lifesat[["Life satisfaction"]].values
```

```python
lifesat.plot(
    kind = "scatter",
    x = "GDP per capita (USD)",
    y = "Life satisfaction",
    grid = True,
)

plt.axis  ([22000, 65000 , 4, 9])

plt.show()
```

```python
print(type(plt))
print(type(lifesat))
print(type(plt.axis))
print(lifesat.columns)
print(hasattr(lifesat, "plot"))
```

```python
model = LinearRegression()
model.fit(x,y)
```

```python
x_new = [[22300,]]
print("prediction:",model.predict(x_new))
```
